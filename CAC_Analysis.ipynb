## Step 1: Import Libraries and Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configure visualization style
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

## Step 2: 2024 Quarterly CAC Data

```python
# 2024 Quarterly CAC Data
data = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "CAC": [225.38, 232.30, 231.35, 229.83],
}

df = pd.DataFrame(data)

# Display the data
print("2024 Quarterly CAC Data:")
print(df)
print()
```

## Step 3: Calculate Key Metrics

```python
# Calculate average CAC
average_cac = round(df["CAC"].mean(), 2)
industry_benchmark = 150
gap = round(average_cac - industry_benchmark, 2)
gap_percent = round((gap / industry_benchmark) * 100, 2)

print(f"Average CAC (2024): ${average_cac}")
print(f"Industry Benchmark: ${industry_benchmark}")
print(f"Gap: ${gap} ({gap_percent}% above benchmark)")
print(f"\nLowest CAC: Q1 at ${df['CAC'].min()}")
print(f"Highest CAC: Q2 at ${df['CAC'].max()}")
print(f"Variability: ${df['CAC'].max() - df['CAC'].min():.2f}")
```

## Step 4: Visualize the Trend

```python
# Create visualization
fig, ax = plt.subplots(figsize=(10, 6))

# Plot quarterly CAC
ax.plot(df["Quarter"], df["CAC"], marker="o", linewidth=2, markersize=8, label="Quarterly CAC", color="#1f77b4")

# Add industry benchmark line
ax.axhline(industry_benchmark, color="red", linestyle="--", linewidth=2, label=f"Industry Target ({industry_benchmark})")

# Annotate each point with CAC value
for i, (quarter, cac) in enumerate(zip(df["Quarter"], df["CAC"])):
    ax.text(i, cac + 1.5, f"${cac:.2f}", ha="center", fontsize=10, fontweight="bold")

# Labels and formatting
ax.set_title("Quarterly Customer Acquisition Cost (CAC) Trend - 2024", fontsize=14, fontweight="bold")
ax.set_xlabel("Quarter", fontsize=12)
ax.set_ylabel("CAC ($)", fontsize=12)
ax.set_ylim(140, max(df["CAC"].max() + 10, industry_benchmark + 10))
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("cac_trend_notebook.png", dpi=200)
plt.show()

print("Visualization saved as 'cac_trend_notebook.png'")
```

## Key Findings

1. **Consistently High CAC**: The average CAC of **$229.72** remains **53% above** the industry benchmark of $150.

2. **Minimal Variation**: CAC fluctuates modestly between $225.38 (Q1) and $232.30 (Q2), indicating a stable but elevated baseline.

3. **Urgency for Optimization**: The persistent gap suggests that current marketing channels and strategies are not cost-efficient and require immediate optimization.

4. **Cost Impact**: If the company acquires 1,000 customers per quarter, the annual CAC overspend is approximately **$79,720** above the industry benchmark.

## Business Implications

- **Margin Compression**: High CAC directly reduces profit margins unless offset by higher customer lifetime value (LTV).
- **Competitive Disadvantage**: Competitors with lower CAC can offer more aggressive pricing or invest more in customer retention.
- **Growth Limitation**: Budget constraints for acquisition may limit market expansion plans.
- **Strategic Risk**: If LTV does not justify the CAC, customer acquisition becomes unsustainable.
